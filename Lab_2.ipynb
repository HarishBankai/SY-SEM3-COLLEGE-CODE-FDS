```python
#to increment and analysis the truth value of a compound propositional statement using python
import numpy as np

def implies(p, q):
  return (not p) or   q

print("Truth of p, q, r  and expression pq -> r")
print("p\tq\tr\tp∧q\t(p∧q)->r")


for p in (True, False):
  for q in (True, False):
    for r in (True, False):
      pq = p and q
      expr = implies(pq, r)
      print(f"{p}\t{q}\t{r}\t{pq}\t{expr}")
```
