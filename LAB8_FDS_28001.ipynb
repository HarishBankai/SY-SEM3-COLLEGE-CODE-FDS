```python
#Data Visualization in Python using Matplotlib, Pandas to plot Scatterplot, Barchart, Linechart and Histogram.

import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

data = {
    'Subjects': ['DSA', 'ADEC', 'FDS', 'DM', 'EEM', 'CEP'],
    'DSA': [86, 77, 75, 78, 82, 95],
    'ADEC': [75, 80, 87, 68, 77, 95],
    'FDS': [96, 70, 65, 88, 72, 78],
    'DM': [76, 81, 75, 90, 82, 87],
    'EEM': [66, 80, 75, 60, 92, 95],
    'CEP': [50, 75, 85, 70, 52, 77] ,
    'Mean of Marks' : [82.16, 80.33, 78.16, 81.83, 78, 68.16] ,
    'Grade': ['A', 'B', 'B', 'A', 'A', 'A']
}

df = pd.DataFrame(data)
print("Original dataset:")
print(df)

print("\nMean of each subject:")
print(df.mean(numeric_only=True))

numeric_df = df.select_dtypes(include='number')
```

```python
plt.scatter(df['DSA'],df['ADEC'])
plt.xlabel("DSA")
plt.ylabel("ADEC")
plt.title("DSA VS ADEC")
plt.show()
```

```python
plt.bar(df['Subjects'],df['Mean of Marks'])
plt.xlabel("Subjects")
plt.ylabel("Mean of Marks")
plt.title("Subjects VS Mean of Marks")
plt.show()
```

```python
plt.plot(df['DSA'],df['FDS'])
plt.xlabel("DSA")
plt.ylabel("FDS")
plt.title("DSA VS FDS")
plt.show()
```

```python
plt.hist(data, bins=10, edgecolor='skyblue')
plt.xlabel('Subjects')
plt.ylabel('Mean of Marks')
plt.title('Subjects Vs Mean of Marks')
plt.show()
```

```python
#Data Visualization in Python using Seaborn to plot Graphs like Boxplot, Heatmap, Faceting, Pairplot.

sns.histplot(df['DSA'], color="skyblue", label="DSA", kde=True, alpha=0.5)
sns.histplot(df['EEM'], color="red", label="EEM", kde=True, alpha=0.5)

plt.title('Distribution of Marks for DSA and EEM')
plt.xlabel('Marks')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(8, 8))
sns.heatmap(numeric_df.corr(), annot=True, cmap='coolwarm', linewidth=0.5)
plt.title('Correlation Heatmap')
plt.show()
```

```python
#Boxplot
plt.figure(figsize=(10, 10))
sns.boxplot(data=df, x='Grade', y='DSA')
plt.title('Boxplot of Grade by DSA')
plt.xlabel('Grade')
plt.ylabel('DSA')
plt.show()
```
