```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt


data = {
'Year': [2015, 2016, 2017, 2018, 2019, 2020, 2021, 2022, 2023, 2024],
'Sales': [150, 180, 220, 250, 300, 280, 350, 400, 420, 460],
'Profit': [30, 40, 45, 50, 60, 65, 70, 75, 80, 90],
'Customers': [800, 950, 1100, 1200, 1300, 1400, 1600, 1750, 1900, 2100],
'Region': ['North', 'South', 'East', 'West', 'North', 'South', 'East', 'West', 'North', 'South']
}


df = pd.DataFrame(data)
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 6))
sns.boxplot(x='Region', y='Sales', data=df)
plt.title('Boxplot of Sales by Region')
plt.xlabel('Region')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()

plt.figure(figsize=(6, 5))
corr = df[['Sales', 'Profit', 'Customers']].corr()
sns.heatmap(corr, annot=True, fmt=".2f", square=True, cbar_kws={"shrink": .8})
plt.title('Heatmap of Correlations')
plt.tight_layout()
plt.show()

sns.pairplot(df[['Sales', 'Profit', 'Customers']], kind='reg', diag_kind='kde')
plt.suptitle('Pairplot of Sales, Profit and Customers', y=1.02)
plt.show()

sns.FacetGrid(df, col='Region', height=4, aspect=1).map(sns.scatterplot, 'Sales', 'Profit').add_legend()
plt.suptitle('Scatterplots of Sales vs Profit by Region', y=1.05)
plt.show()
```
