```python
import numpy as np

#declare array using np.array
A = np.array([[4, 6],
              [6, 9]])

B = np.array([[3, 4] ,
             [7, 9]])

print("First Matrix : \n", A)  #print the array
print("Second Matrix : \n", B)

rank = np.linalg.matrix_rank(A)  #use rank syntax for formula

determinant = np.linalg.det(A)

transpose = A.T

if(np.linalg.det(A) != 0):
  inverse = np.linalg.inv(A)
  print("\nInverse is \n", inverse)
else:
  print("\nInverse does not exist \n")

eigenvalues, eigenvectors = np.linalg.eig(A)

result = np.dot(A, B)

print("Rank is \n", rank)
print("\nDeterminant is \n", determinant)
print("\nTranspose is \n", transpose)
print("\nEigenvalues are \n", eigenvalues)
print("\nEigenvectors are \n", eigenvectors)

print("Result of Multiplication Matrix : \n", result)
```
