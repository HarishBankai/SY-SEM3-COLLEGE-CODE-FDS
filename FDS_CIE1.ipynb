```python
import numpy as np

A = np.array([[3, 2] ,
              [4, 5]])

print("Matrix : " , A)

rank = np.linalg.matrix_rank(A)
print("Rank : " , rank)

determinant = np.linalg.det(A)
print("Determinant : " , determinant)

if(determinant != 0):
    inverse = np.linalg.inv(A)
    print("Inverse : " , inverse)
else:
    print("Invalid.(Det = 0)")

eigenvalues , eigenvectors = np.linalg.eig(A)
print("Eigenvalues : " , eigenvalues)
print("Eigenvectors : " , eigenvectors)

transpose = A.T
print("Transpose : " , transpose)
```

```python
import numpy as np

A = np.array([[3, 2] ,
              [4, 5]])

B = np.array([[1, 2] ,
              [6, 7]])

product = np.dot(A, B)

print("Matrix : " , product)
```

```python
import pandas as pd

patients = [
    {"Name": "Alice", "Weight": 55, "Height": 1.60},
    {"Name": "Bob", "Weight": 75, "Height": 1.75},
    {"Name": "Charlie", "Weight": 68, "Height": 1.70},
]

df = pd.DataFrame(patients)
print(df)

bmi = df["Weight"] / (df["Height"] ** 2)
df["BMI"] = bmi

print(df)
```
