```python
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, LabelEncoder

# Define a dictionary containing data for 4 employees
data = {
    'Date': ['2023-01-10', '2023-03-15', '2023-02-20', '2023-01-25'],
    'Age': [25, 45, 35, 22],
    'Salary': [50000, 80000, 60000, 45000],
    'Department': ['HR', 'Finance', 'IT', 'HR'],
    'PerformanceScore': [3.5, 4.0, 3.0, 2.5]
}

# Create a DataFrame from the dictionary
df = pd.DataFrame(data)

# Convert date to datetime format
df['Date'] = pd.to_datetime(df['Date'])

# Initialize MinMaxScaler and scale 'Age' and 'Salary' columns
scaler = MinMaxScaler()
df[['Age', 'Salary']] = scaler.fit_transform(df[['Age', 'Salary']])

# Define bins for performance scores, labels for bins and create a new column 'PerformanceCategory'
bins = [0, 2.9, 3.5, 4.5]
labels = ['Low', 'Medium', 'High']
df['PerformanceCategory'] = pd.cut(df['PerformanceScore'], bins=bins, labels=labels)

# Initialize LabelEncoder and transform the 'Department' column to numerical values
le = LabelEncoder()
df['Department_Label'] = le.fit_transform(df['Department'])

# Convert 'Department' into dummy variables using get_dummies function
df = pd.get_dummies(df, columns=['Department'], prefix='Dept')

print(df)  # Print the DataFrame
```
