```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load dataset (download from Kaggle or use a local CSV)
df = pd.read_csv("/content/StudentsPerformance.csv")

# Display first few rows
print(df.head())

# Focus on 'math score' column
sns.boxplot(x=df['math score'], color='lightgreen')
plt.title('"Boxplot of Math Scores')
plt.show()

# Find outliers using IQR
Q1 = df['math score'].quantile(0.25)
Q3 = df['math score'].quantile(0.75)
IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = df[(df['math score'] < lower_limit) | (df['math score'] > upper_limit)]
print("\nOutliers in Math Scores:")
print(outliers[['gender', 'math score']])  # fix this code
```
